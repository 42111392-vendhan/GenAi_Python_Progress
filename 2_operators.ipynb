### **Operators**

## *arithmetic operator*

```python
print(10 + 3) # 13
print(10-3) #  7
print(10 * 3) # 30
print(10 / 3) # 3.333....
print(10 ** 3)# pwo = 10*10*10 = 1000
print(10 // 3)# Q => 3
print(10 % 3) # R => 1
```

## *assignment operator*

```python
a = 10
print(a)
a += 5 
print(a)
a *= 2 
print(a)
b = 0
b += 20
print(b)
```

## *comparison operator*

```python
print(10 == 10) 
print(10 != 10) 
print(10 > 10) 
print(10 >= 10) 
print(10 < 10) 
print(10 <= 10)
```

## *logical operator*

```python
print('----AND----')
print(10 == 10 and 6 > 4) 
print(10 != 10 and 6 > 4) 
print(10 == 10 and 6 < 4) 
print(10 != 10 and 6 < 4) 
print('----OR----')
print(10 == 10 or 6 > 4) 
print(10 != 10 or 6 > 4) 
print(10 == 10 or 6 < 4) 
print(10 != 10 or 6 < 4) 
print('----Not----')
print(not 10==10) 
print(not 10!=10) 
```

### **Python Collection**
#### **List**

*is a collection which is ordered and changeable.
Allows duplicate members. Denoted by []*

```python
mylist = [10,20.55,"Hello",True,10,"Hello",20.55,False,True,10,20.55,True]
print(mylist,type(mylist))
print(mylist[5])
print(mylist[8])
mylist[8]="Welcome"
print(mylist)
```

#### **Tuple**

*is a collection which is ordered and unchangeable.
Allows duplicate members. Denoted by()*

```python
mytuple = (10,20.55,"Hello",True,10,"Hello",20.55,False,True,10,20.55,True)
print(mytuple,type(mytuple))
print(mytuple[4])
print(mytuple[7])
mytuple[7]="Python"
print(mytuple)
```

#### **Set**

*is a collection which is unordered and unindexed.
No duplicate members. Denoted by{}*

```python
myset = {100,"hai",100,"hello","hai",50.55,100,"hello",50.55,True,False}
print(myset,type(myset))
```

#### **Dictionary**
*is a collection that is unordered, changeable (mutable), and stores
data in key-value pairs. Keys must be unique (no duplicate keys). It is denoted
by curly braces {} using the format{key: value}*

```python
mydict = {
"Name":"Jeevan",
"Course":"Python",
"Age":15,
100:50.55,
40.55:True,
}
print(mydict,type(mydict))

print(mydict["Name"])
mydict["Age"]= 16
mydict["City"]="Chennai"
print(mydict)
```

### **Mutable vs Immutable Objects**
**Immutable** *(int, f loat, bool, str, tuple)*
* *Same data stores in Same location*
* *Different Data stores in Different location*

```python
a = 100
b = 100
c = 101
print(a,type(a),id(a))
print(b,type(b),id(b))
print(c,type(c),id(c))
```

```python
x = (10,20,30,40,50)
y = (10,20,30,40,50)
print(x,type(x),id(x))
print(y,type(y),id(y))
```

#### **Mutable**
* *Same data stores in Different location*
* *Different Data stores in Different location*

```python
x = [10,20,30,40,50]
y = [10,20,30,40,50]
print(x,type(x),id(x))
print(y,type(y),id(y))
```

```python
a = 100
b = 100
print(a != b) 
print(a is not b)
```

```python
x = (10,20,30,40,50)
y = (10,20,30,40,50)
print(x != y)
print( x is y)
```
