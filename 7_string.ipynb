```python
def star_pattern(n):
    for x in range(n, 0, -1):
        for y in range(x):
            print("#",end=" ")
        print()
star_pattern(5)
```

### **String formatting**
* %d = integer
* %s = string
* %f = float with default 6 decimal places
* %.2f = float with two decimal places
* %.10f = float with ten decimal places


```python
name = "John Doe"
age = 30
height = 1.75
print("%s is %d year old and height is %.3f" %(name, age, height))
```
