### **Interview**
*Interview   check palindrome*

```python
mystr = "racecar"
revstr = mystr[::-1]
print(mystr)
print(revstr)
if mystr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = "MalaYa lAm"
newstr = mystr.lower().replace(" ",'')
revstr = newstr[::-1]
print(mystr)
print(newstr)
print(revstr)
if newstr == revstr:
    print(True)
else:
    print(False)
```

```python
mystr = "Sir, I demand, I am a maid named Iris"
newstr = mystr.lower().replace(" ",'').replace(',','')
revstr = newstr[::-1]
print(mystr)
print(newstr)
print(revstr)
if newstr == revstr:
    print(True)
else:
    print(False)
```

```python
def Check_palindrome(mystr):
    newstr = ''
    for each in mystr:
        if each.isalpha():
            newstr += each.lower() # newstr += each.lower()
    if newstr == newstr[::-1]:
        return True
    else:
        return False
ans = Check_palindrome("MalaYa  lAm")
print(ans)
print(Check_palindrome("racecar"))
print(Check_palindrome("Red roses run no risk. Sir, on Nurse's Order. "))
print(Check_palindrome("Sir, I demand, I am a maid named Iris"))
```
