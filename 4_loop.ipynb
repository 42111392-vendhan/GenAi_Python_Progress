### **Python While loop** *(condition statements)*
*Syntax*
* *while \<condition>:*

```python
a = 1 
while a<=10: 
    print(a,end=" ")
    a += 1 
print("\nLoop Over")
```

```python
a = 0
while a<=100:
    print(a,end=" ")
    a += 10
print("\nLoop Over")
```

```python
a = 100
while a >= 0:
    print(a, end = " ")
    a-= 10
```

### **break** *–stop the loop*

```python
a = 1
while a>0:
    if a == 5:
        print(a)
        break
    print(a,end=" ")
    a+=1
```

### **continue** *– go back to the top of my loop*

```python
a = 1
while a<=20:
    if a % 3 == 0:
        print(a,end=" ")
        a += 1
        continue
    a += 1
print("\nLoop over")
```

### **Nested while Loop**

```python
a = 1
while a<=5:
    b = 1
    while b<=5:
        print(b,end="")
        b += 1
    a += 1
    print()
```

```python
a = 1
while a<=3:
    b = 1
    while b<=3:
        print(b,end="")
        b += 1
    a += 1
    print()
```

```python
row = int(input("Enter a roe Number"))
column = int(input("Enter a coumn Number"))
a = 1
while a<=row:
    b = 1
    while b<=column:
        print(b,end="")
        b +=1
    a +=1
    print()
```

### **Python for Loop**
*Syntax*

**for \<variable_name> in \<str/list/tuple/set/dict>:**

```python
mystr = "Hello world!"
for x in mystr: 
    print(x,end="")
```

```python

mylist = [10,20.55,"hello",True,20.55,"hello"]
for x in mylist:
    print(x,end =",")
```

**range()** *–is a python standard function used for numbers*

```python
for x in range(10+1):
    print(x,end=' ')
```

### **for with else**

```python
for x in "Hello world":
    print(x,end = "")
else:
    print("\nThe above loop complieted successfully")
```

### **Nested Loop**

```python
for x in range(5):
    for y in range(1,5+1):
        print(y,end='')
    print()
```

```python
for x in ["Yellow","Green","Red"]:
    print("----------------------")
    for y in ["Cat","Dog","Car","House"]:
        print(x,y)
    print("----------------------")
print("Loop Over")
```
