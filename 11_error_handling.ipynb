### **Handling a Exception**


```python
print("Few lines above")
try:
    num = int(input("Enter a number"))
    print(f"The user entered number is = {num}")
except:
    print("Error handling is done here")
print("Few lines below")
```

### **Getting an input from user with Error Handling.**


```python
print("Few lines above")
while True:
    try:
        num = int(input("Enter a number"))
        print(f"The user entered number is = {num}")
        break
    except:
        print("Please Enter a Integer Number eg, -99, 100, -662345")
print("Few lines below")
```

### **Common Exception Base class to store the error message**


```python
print("Few lines above")
while True:
    try:
        num1 = int(input("Enter a number"))
        print(f"The User Entered Number is = {num1}")
        break
    except Exception as e:
        print(e)
print("Few lines below")
```

### **Dividing a number with Error Handling.**


```python
print("Few lines above")
try:
    num1 = int(input("Enter a Dividend number"))
    print(f"The User Entered Dividend Number is = {num1}")
    num2 = int(input("Enter a Divisor number"))
    print(f"The User Entered Divisor Number is = {num2}")
    quotient = num1/num2
    print(f"The Quotient of {num1}/{num2} = {quotient}")
except ValueError:
    print("You gave wrong input type")
except ZeroDivisionError:
    print("Dont divide by zero")
print("Few lines below")
```

### **The **else** block works only if try is correct**


```python
print("Few lines above")
try:
    num1 = int(input("Enter a number"))
    print(f"The User Entered Number is = {num1}")
except:
    print("Error handling is done here")
else:
    print("Executes since try has no errors")
    print("Few lines below")
```

### **Error Handling with Raise**


```python
num = int(input("Enter a number inbetween 10-50:"))
if num<10:
    raise ValueError
elif num>50:
    raise ZeroDivisionError
else:
    print(f"Thank you for entering the = {num}")
```

### **Handling error with raise & Except**


```python
while True:
    try:
        num = int(input("Enter a number inbetween 10-50:"))
        if num<=10:
            raise ValueError
        elif num>=50:
            raise ZeroDivisionError
        else:
            print(f"Thank you for entering the {num}")
            break
    except ValueError:
        print("Please Enter a numbaer greater than or equal to 10")
    except ZeroDivisionError:
        print("Please Enter a number lesser than or equal to 50")
```

### **Handling error with raise, Except & as**


```python
while True:
    try:
        num = int(input("Enter a number inbetween 10-50:"))
        if num<10:
            raise ValueError("Please Enter a number greater than or equal to 10")
        elif num>50:
            raise ZeroDivisionError("Please Enter a number lesser than or equal to 50")
        else:
            print(f"Thank you for entering the = {num}")
            break
    except ValueError as x:
        print(x)
    except ZeroDivisionError as y:
        print(y)
```

### **assert - keyword**

```python
while True:
    try:
        num = int(input("Enter a number between 10-50:"))
        assert num >= 10 and num <= 50, "Please provide a number inbetween 10-50"
        print(f"Thank you for entering the = {num}")
        break
    except Exception as e:
        print(e)
```

### **What is Nested Error Handling in Python?**


```python
try:
    print("Outer try block")
    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print(result)
    except ValueError:
        print("Inner Error: Invalid input (not a number)")
    except ZeroDivisionError:
        print("Inner Error: Cannot divide by zero")
except Exception as e:
    print("Outer Error:", e)
```
