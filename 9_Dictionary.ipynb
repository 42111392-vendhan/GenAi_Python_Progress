### **Dictionary functions**

```python
mydict = {
    "name": "John",
    "age": 30,
    "city": "New York"
}
print(mydict)
```

```python
mydict = {
"Name":"Honey",
"Age" : 35,
"City": "Cennai",
500 : 50.55,
}
print(mydict,type(mydict))
print(mydict[500])
mydict["City"]='Mumbai'
mydict["Course"]="Python"
print(mydict)
```

*Keys()*

```python
mydict = {
"Name":"Honey",
"Age" : 35,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
print()
print(list(mydict.keys()))
print()
ans = list(mydict.keys())
print(ans[1])
print()
for each in mydict.keys():
    print(each)
```

*Values()*

```python
mydict = {
"Name":"Honey",
"Age" : 35,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
print(list(mydict.values()))
print()
for each in mydict.values():
    print(each)
```

*Items()*

```python
mydict = {
"Name":"Honey",
"Age" : 35,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
print()
print(list(mydict.items()))
print()
for each in mydict.items():
    print(f"{each[0]}\t{each[1]}")
```

*get()*

```python
mydict = {
"Name":"Honey",
"Age" : 35,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
print(mydict[500])
print(mydict[600])
print("Hello Python")
```

```python
mydict = {
"Name":"Honey",
"Age" : 35,
"City": "Cennai",
500 : 50.55,
}
print(mydict)
print(mydict["Name"])
print(mydict.get(500))
print(mydict.get(600))
print('Hello Python')
```

*copy()*

```python
x = {"A":100,"B":200,"C":300,}
y = {"A":100,"B":200,"C":300,}
x["D"]=400
y["E"]=500
print(f"X = {x},{type(x)},{id(x)}")
print(f"Y = {y},{type(y)},{id(y)}")
```

```python
x = {"A":100,"B":200,"C":300,}
y = x.copy()
x["D"]=400
y["E"]=500
print(f"X = {x},{type(x)},{id(x)}")
print(f"Y = {y},{type(y)},{id(y)}")
```

### **Nested Dictionary**

```python
mydict = {
"Name": "Manjunathan",
"City":["Chennai","Kovai","Mumbai"],
"Year":(2000,2010,2020)
}
print(mydict,type(mydict))
print(mydict.keys())
print(mydict.get("Name"))
print(mydict.get("City"))
print(mydict.get("City")[0])
print(mydict.get("City")[1])
print("-----------------------")
print(mydict.get("Year"))
print(mydict.get("Year")[1])
print("-----------------------")
print(mydict.get("City")[2][3:])
```

```python
import pprint
x = {
    "menu": {
            "id": "file",
            "value": "File",
            "popup": {
                    "menuitem": [
                                {"value": "New", "onclick": "CreateNewDoc()"},
                                {"value": "Open", "onclick": "OpenDoc()"},
                                {"value": "Close", "onclick": "CloseDoc()"}
                                ]
                    }
            }
    }
print(x,type(x))
print(x.keys())
print(x.get("menu"))
print(x.get("menu")["popup"])
print(x.get("menu")["popup"]["menuitem"])
print('------------------')
print(x.get("menu")["popup"]["menuitem"][0])
print(x.get("menu")["popup"]["menuitem"][1])
print(x.get("menu")["popup"]["menuitem"][2])
print('------------------')
print(x.get("menu")["popup"]["menuitem"][0]["onclick"])
print(x.get("menu")["popup"]["menuitem"][2]["onclick"])
```
